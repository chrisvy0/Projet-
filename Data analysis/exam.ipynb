# PARTIE	3	:	ANALYSE	EXPLORATOIRE		

## Airbnb : 
• histogramme des prix 
• barplot prix moyen par quartier 

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Histogramme simple
plt.figure(figsize=(10,6))
sns.histplot(df1['price'], bins=30, kde=True, color='skyblue')
plt.title("Distribution des prix des logements Airbnb")
plt.xlabel("Prix")
plt.ylabel("Nombre de logements")
plt.show()
```

```python
# Calcul du prix moyen par quartier
prix_moyen_quartier = df1.groupby('neighbourhood')['price'].mean().sort_values(ascending=False)

# Barplot
plt.figure(figsize=(12,6))
sns.barplot(x=prix_moyen_quartier.index, y=prix_moyen_quartier.values, palette='viridis')
plt.title("Prix moyen par quartier")
plt.xlabel("Quartier")
plt.ylabel("Prix moyen")
plt.xticks(rotation=45)  # rotation des labels pour plus de lisibilité
plt.show()
```

## Taxi : 
• scatter plot distance vs prix 
• histogramme durée trajet 

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df3, x='distance_km', y='fare_amount', hue='pickup_zone', palette='Set2')
plt.title("Relation entre distance et tarif des trajets")
plt.xlabel("Distance (km)")
plt.ylabel("Tarif (USD)")
plt.legend(title="Pickup zone")
plt.show()
```

```python
# Calcul du chiffre d'affaires total par catégorie
ventes_par_categorie = df2.groupby('product_category')['total'].sum().reset_index()

# Barplot
plt.figure(figsize=(8,5))
sns.barplot(x='product_category', y='total', data=ventes_par_categorie, palette='viridis')
plt.title("Chiffre d'affaires par catégorie")
plt.ylabel("Total ventes")
plt.xlabel("Catégorie")
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df2['total'], bins=20, kde=True, color='skyblue')
plt.title("Distribution du montant total des transactions")
plt.xlabel("Montant total")
plt.ylabel("Nombre de transactions")
plt.show()
```
